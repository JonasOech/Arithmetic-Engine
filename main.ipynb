````python
from anytree import PreOrderIter
from IPython.display import Markdown
from anytree.exporter import MermaidExporter
from parser import Parser
from transformation import Transformation
from random import shuffle
````

````python
transformations = {
    "Multiplicative identity":    Transformation("a * 1 = a"),
    "Multiplicative zero":        Transformation("a * 0 = 0"),
    "Exponent zero":              Transformation("a ^ 0 = 1"),
    "Distributive law":           Transformation("a * ( b + c ) = a * b + a * c"),
    "Binomial square (sum)":      Transformation("( a + b ) ^ 2 = a ^ 2 + 2 a b + b ^ 2"),
    "Binomial square (diff)":     Transformation("( a - b ) ^ 2 = a ^ 2 - 2 a b + b ^ 2"),
    "Difference of squares":      Transformation("a ^ 2 - b ^ 2 = ( a + b ) * ( a - b )"),
    "Log of product":             Transformation("log ( a * b ) = log ( a ) + log ( b )"),
    "Power of product":           Transformation("( a * b ) ^ c = a ^ c * b ^ c"),
    "Exponent addition":          Transformation("a ^ b * a ^ c = a ^ ( b + c )"),
}
````

````python
parser = Parser()
tree = parser.parse_from_txt("a + b * c")
transformationItems = [t[1] for t in list(transformations.items())]
for _ in range(10):
    shuffle(transformationItems)
    for transform in transformationItems:
        for formula in PreOrderIter(tree):
            out, tree = transform.apply(tree)
            if out == True:
                print(f"{parser.parse_to_txt(tree)} applied {str(transform)}")
                break

        if out == True: 
            break
````

````python

mermaid_text = "\n".join(MermaidExporter(tree))
display(Markdown(f"```mermaid\n{mermaid_text}\n```"))
````
