````python
from anytree import RenderTree
from IPython.display import Markdown
from anytree.exporter import MermaidExporter
from parser import Parser
from transformation import Transformation
````

````python
tree = Parser().parse_from_txt(" ( a + b ) ^ 2")
binom = Transformation(" ( a + b ) ^ 2 = a ^ 2 + 2 a b + b ^ 2")

out, tree = binom.apply(tree)

mermaid_text = "\n".join(MermaidExporter(tree))
display(Markdown(f"```mermaid\n{mermaid_text}\n```"))
````
