````python
from anytree import RenderTree
from IPython.display import Markdown
from anytree.exporter import MermaidExporter
from parser import Parser
````

````python
parser = Parser()
tree = parser.parse_from_txt("2 ^ 30 ( x1 + 3 ) ^ 2 - sin ( x1 ) ^ 3 ^ 4 ")
mermaid_text = "\n".join(MermaidExporter(tree))
display(Markdown(f"```mermaid\n{mermaid_text}\n```"))
````
